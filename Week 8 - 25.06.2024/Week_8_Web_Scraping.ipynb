```python
# Starting with some basics.
import requests
from bs4 import BeautifulSoup

url = 'https://www.crummy.com/software/BeautifulSoup/bs4/doc/'

# You probably will remember this step from our previous weeks. We imported the Ricardo and Adam Smith this way as well.
response = requests.get(url)

# There are several responses a server can give you, most common ones are 200, 404 and 502.
if response.status_code == 200:
    # Parse the content of the response using lxml
    soup = BeautifulSoup(response.content, 'lxml')

    # Find elements by tag, here we look for all 'a' tags (hyperlinks)
    links = soup.find_all('h1')

    # Print the href attribute of each link
    for link in links:
        print(link.text)
else:
    print(f"Failed to retrieve the webpage: Status code {response.status_code}")
```

```python
paragraphs = soup.find_all('p')
for p in paragraphs:
    print(p.get_text())
```

```python
# Find the table by an ID or class
table = soup.find('table')
rows = table.find_all('tr')

for row in rows:
    columns = row.find_all('td')
    # Extract text from each column
    column_data = [col.get_text() for col in columns]
    print(column_data)
```

```python
tables = soup.find_all('table')

for index, table in enumerate(tables):
    rows = table.find_all('tr')
    data = [[td.get_text().strip() for td in tr.find_all('td')] for tr in rows if tr.find_all('td')]

    # If there are headers in the first row
    headers = [th.get_text().strip() for th in rows[0].find_all('th')] if rows[0].find_all('th') else []

    df = pd.DataFrame(data, columns=headers if headers else None)
```

```python
# set the target URL
url = "https://www.scrapethissite.com/pages/"

# set up the webdriver
driver = webdriver.Chrome(options=chrome_options)
```

```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
```

```python
driver.get("https://www.scrapethissite.com/pages/")
countries_link = driver.find_element(By.XPATH, "//a[contains(@href, '/pages/simple/')]")
countries_link.click()

soup = BeautifulSoup(driver.page_source, 'html.parser')
countries = soup.find_all('div', class_='col-md-4 country')

countries_list = []
capitals_list = []
populations_list = []
areas_list = []

for country in countries:
    country_name = country.find('h3', class_='country-name').get_text(strip=True)
    capital = country.find('span', class_='country-capital').get_text(strip=True)
    population = country.find('span', class_='country-population').get_text(strip=True)
    area = country.find('span', class_='country-area').get_text(strip=True)

    countries_list.append(country_name)
    capitals_list.append(capital)
    populations_list.append(population)
    areas_list.append(area)

df = pd.DataFrame({
    'Country': countries_list,
    'Capital': capitals_list,
    'Population': populations_list,
    'Area (km^2)': areas_list
})
```
